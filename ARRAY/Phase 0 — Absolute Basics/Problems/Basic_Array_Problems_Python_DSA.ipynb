# Basic Array Problems — Python DSA

This notebook covers four foundational array problems:

1. Find largest element
2. Find smallest element
3. Find second largest element
4. Find second smallest element



## 1. Find Largest Element

### Problem
Given an array, find the largest element without sorting the array.

### Approach
Initialize `largest` with the first element. Traverse the array and update it whenever a larger value is encountered.

### Complexity
- **Time:** O(n)
- **Extra Space:** O(1)


```python
arr = [10, 25, 7, 42, 18]

largest = arr[0]

for x in arr:
    if x > largest:
        largest = x

print("Largest:", largest)
```

## 2. Find Smallest Element

### Problem
Given an array, find the smallest element without sorting the array.

### Approach
Initialize `smallest` with the first element. Traverse the array and update it whenever a smaller value is encountered.

### Complexity
- **Time:** O(n)
- **Extra Space:** O(1)


```python
arr = [10, 25, 7, 42, 18]

smallest = arr[0]

for x in arr:
    if x < smallest:
        smallest = x

print("Smallest:", smallest)
```

## 3. Find Second Largest Element

### Problem
Find the **second distinct largest** element in the array.

### Approach
Maintain two variables:
- `largest`
- `second_largest`

During one traversal:
- If the current value is greater than `largest`, move the old `largest` into `second_largest`.
- Otherwise, if the value lies between them and differs from `largest`, update `second_largest`.

### Complexity
- **Time:** O(n)
- **Extra Space:** O(1)

This avoids sorting, so it is more efficient than an O(n log n) sorting-based solution.


```python
arr = [10, 25, 7, 42, 18]

largest = float('-inf')
second_largest = float('-inf')

for x in arr:
    if x > largest:
        second_largest = largest
        largest = x
    elif x > second_largest and x != largest:
        second_largest = x

print("Second Largest:", second_largest)
```

## 4. Find Second Smallest Element

### Problem
Find the **second distinct smallest** element in the array.

### Approach
Maintain two variables:
- `smallest`
- `second_smallest`

During one traversal:
- If the current value is smaller than `smallest`, move the old `smallest` into `second_smallest`.
- Otherwise, if the value lies between them and differs from `smallest`, update `second_smallest`.

### Complexity
- **Time:** O(n)
- **Extra Space:** O(1)

This is a one-pass solution and does not require sorting.


```python
arr = [10, 25, 7, 42, 18]

smallest = float('inf')
second_smallest = float('inf')

for x in arr:
    if x < smallest:
        second_smallest = smallest
        smallest = x
    elif x < second_smallest and x != smallest:
        second_smallest = x

print("Second Smallest:", second_smallest)
```

## Complexity Summary

| Problem | Time | Extra Space |
|---|---:|---:|
| Largest element | O(n) | O(1) |
| Smallest element | O(n) | O(1) |
| Second largest | O(n) | O(1) |
| Second smallest | O(n) | O(1) |

## Key DSA Pattern

The main pattern to remember is maintaining two candidates during a single traversal:

- `largest` + `second_largest`
- `smallest` + `second_smallest`

These are foundational problems before moving to duplicates, rotations, missing numbers, two pointers, prefix sums, and other array patterns.
