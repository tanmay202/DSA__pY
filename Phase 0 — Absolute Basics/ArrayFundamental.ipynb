# Array Fundamentals


## 1. Array Creation and Indexing

In Python, arrays for most DSA problems are represented using the built-in `list` type.

Python lists use **zero-based indexing**, so the first element is at index `0`. Negative indexes access elements from the end.

Important concepts:
- Accessing an element by index is **O(1)**.
- Updating an element by index is **O(1)**.
- A list containing `n` elements requires **O(n)** storage.

The following code demonstrates creation, positive indexing, negative indexing, and updating an element.


````python
arr = [10, 20, 30, 40, 50]

# Indexing
print(arr[0])   # 10
print(arr[1])   # 20
print(arr[4])   # 50

# Negative indexing
print(arr[-1])  # 50
print(arr[-2])  # 40

# Modify an element
arr[2] = 100

print(arr)
# [10, 20, 100, 40, 50]
````

## 2. Array Traversal 

Traversal means visiting every element of an array.

For DSA, you should be comfortable with:
- Direct value traversal
- Index-based traversal
- Traversal using `enumerate()` when both the index and value are needed

For an array of `n` elements, a complete traversal takes **O(n)** time and **O(1)** extra space when no additional data structure is created.


````python
arr = [10, 20, 30, 40, 50]

# Direct traversal
for x in arr:
    print(x)

# Index-based traversal
for i in range(len(arr)):
    print(arr[i])

# Index + value
for i, value in enumerate(arr):
    print(i, value)
````

## 3. Insert / Delete

Python lists are dynamic arrays.

Common operations:
- `append()` adds an element at the end and is **O(1) amortized**.
- `insert()` at the beginning or middle is **O(n)** because elements may need to shift.
- `pop()` without an index removes the last element and is typically **O(1)**.
- `pop(index)` and `remove(value)` can be **O(n)** because elements may need to shift or the list must be searched.

Understanding these costs is important for choosing the right data structure in DSA.


````python
# Insert at end
arr = [10, 20, 30]
arr.append(40)
print(arr)
# [10, 20, 30, 40]

# Insert at beginning
arr = [10, 20, 30]
arr.insert(0, 5)
print(arr)
# [5, 10, 20, 30]

# Insert at a position
arr = [10, 20, 30, 40]
arr.insert(2, 99)
print(arr)
# [10, 20, 99, 30, 40]

# Delete last element
arr = [10, 20, 30]
arr.pop()
print(arr)
# [10, 20]

# Delete by index
arr = [10, 20, 30, 40]
arr.pop(1)
print(arr)
# [10, 30, 40]

# Delete by value
arr = [10, 20, 30, 20]
arr.remove(20)
print(arr)
# [10, 30, 20]
````

## 4. Reverse an Array

Reversing an array is a classic foundational problem.

Three useful approaches are shown:
1. Slicing — simple, but creates a new list.
2. `reverse()` — in-place.
3. Two pointers — the core DSA technique.

The two-pointer technique uses a `left` pointer at the beginning and a `right` pointer at the end, swapping values while moving toward the center.

For the in-place two-pointer approach:
- Time: **O(n)**
- Extra space: **O(1)**


````python
# Method 1 — slicing
arr = [1, 2, 3, 4, 5]

arr = arr[::-1]

print(arr)
# [5, 4, 3, 2, 1]

# Method 2 — .reverse()
arr = [1, 2, 3, 4, 5]

arr.reverse()

print(arr)

# Method 3 — Two pointers
arr = [1, 2, 3, 4, 5]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]

    left += 1
    right -= 1

print(arr)
````

## 5. Find Minimum / Maximum

Finding the minimum or maximum is a simple but important single-pass array problem.

The DSA implementation keeps:
- a `minimum` variable for the smallest value seen so far
- a `maximum` variable for the largest value seen so far

A single traversal can compute both values efficiently:
- Time: **O(n)**
- Extra space: **O(1)**

Python also provides the built-in `min()` and `max()` functions, which also require linear traversal of the array.


````python
arr = [10, 4, 25, 7, 2]

# Built-in approach
print(min(arr))  # 2
print(max(arr))  # 25

# DSA implementation — minimum
minimum = arr[0]

for x in arr:
    if x < minimum:
        minimum = x

print(minimum)

# DSA implementation — maximum
maximum = arr[0]

for x in arr:
    if x > maximum:
        maximum = x

print(maximum)

# Find both simultaneously
minimum = arr[0]
maximum = arr[0]

for x in arr:
    if x < minimum:
        minimum = x

    if x > maximum:
        maximum = x

print("Min:", minimum)
print("Max:", maximum)
````

## 6. Sum / Product


Aggregate operations reduce an array to a single value.

For sum:
- Start from `0`
- Add each element

For product:
- Start from `1`
- Multiply each element

Both require one complete traversal:
- Time: **O(n)**
- Extra space: **O(1)**

Python also provides `sum(arr)` for summation.


````python
arr = [1, 2, 3, 4, 5]

# Sum
total = 0

for x in arr:
    total += x

print(total)
# 15

# Product
product = 1

for x in arr:
    product *= x

print(product)
# 120

# Python shortcut for sum
print(sum(arr))
````

## 7. Frequency Counting

Frequency counting is one of the most important foundations for DSA.

A dictionary maps each value to the number of times it occurs.

Example:

`[1, 2, 2, 3, 1, 2, 4]`

produces:

`{1: 2, 2: 3, 3: 1, 4: 1}`

Average complexity:
- Time: **O(n)**
- Extra space: **O(k)**, where `k` is the number of unique elements.

This pattern becomes extremely useful in hashing, anagrams, duplicates, top-k problems, and many array/string interview questions.


````python
arr = [1, 2, 2, 3, 1, 2, 4]

# Explicit dictionary approach
freq = {}

for x in arr:
    if x in freq:
        freq[x] += 1
    else:
        freq[x] = 1

print(freq)

# Compact Python approach
freq = {}

for x in arr:
    freq[x] = freq.get(x, 0) + 1

print(freq)
````

## 8. Searching

Searching asks whether a target value exists and, commonly, where it occurs.

### Linear Search
Works on both sorted and unsorted arrays.
- Best: **O(1)**
- Average: **O(n)**
- Worst: **O(n)**
- Extra space: **O(1)**

### Binary Search
Requires the array to be sorted.
It repeatedly cuts the search space approximately in half.
- Time: **O(log n)**
- Extra space: **O(1)** for the iterative implementation.

You should be able to implement both without using a library search function.


````python
# Linear Search
arr = [10, 20, 30, 40, 50]
target = 30

found = False

for i in range(len(arr)):
    if arr[i] == target:
        print("Found at index", i)
        found = True
        break

if not found:
    print("Not found")


# Linear Search — function version
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i

    return -1


print(linear_search([10, 20, 30, 40], 30))


# Binary Search
arr = [10, 20, 30, 40, 50, 60]
target = 40

left = 0
right = len(arr) - 1

while left <= right:

    mid = (left + right) // 2

    if arr[mid] == target:
        print("Found at", mid)
        break

    elif arr[mid] < target:
        left = mid + 1

    else:
        right = mid - 1
````

## 9. Copying Arrays

Python variable assignment does **not** create a new list.

When you write:

`b = a`

both variables reference the same list, so changing `b` also changes `a`.

To create a separate shallow copy, use:
- slicing: `a[:]`
- `.copy()`

Copying `n` elements requires:
- Time: **O(n)**
- Space: **O(n)**

This distinction between references and copies is important when writing reliable DSA code.


````python
# Assignment creates another reference to the same list
a = [1, 2, 3]

b = a

b[0] = 100

print(a)
# [100, 2, 3]


# Shallow copy using slicing
a = [1, 2, 3]

b = a[:]

b[0] = 100

print(a)
print(b)


# Copy using .copy()
a = [1, 2, 3]

b = a.copy()

print(a)
print(b)
````

## 10. 2D Arrays

A 2D array in Python is commonly represented as a **list of lists**.

For example:

```text
1 2 3
4 5 6
7 8 9
```

The element at row `i` and column `j` is accessed using:

`matrix[i][j]`

A full traversal of an `r × c` matrix visits every cell:
- Time: **O(r × c)**
- Extra space: **O(1)** if no additional matrix is created.

The following examples cover creation, indexing, traversal, row sums, column sums, and the main diagonal.


````python
# Create a 2D array
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Access elements
print(matrix[0][0])  # 1
print(matrix[1][2])  # 6
print(matrix[2][1])  # 8


# Traverse 2D array using indexes
for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        print(matrix[i][j])


# Cleaner traversal
for row in matrix:
    for value in row:
        print(value)


# Row sums
for row in matrix:
    total = 0

    for x in row:
        total += x

    print(total)


# Column sums
rows = len(matrix)
cols = len(matrix[0])

for j in range(cols):

    total = 0

    for i in range(rows):
        total += matrix[i][j]

    print(total)


# Main diagonal
n = len(matrix)

for i in range(n):
    print(matrix[i][i])
````

## 11. Time and Space Complexity

Big-O notation describes how an algorithm scales as input size grows.

### Common array complexities

| Operation | Time | Extra Space |
|---|---:|---:|
| Access `arr[i]` | O(1) | O(1) |
| Update | O(1) | O(1) |
| Traversal | O(n) | O(1) |
| Linear Search | O(n) worst case | O(1) |
| Binary Search | O(log n) | O(1) |
| Append | O(1) amortized | O(1) |
| Insert beginning | O(n) | O(1) |
| Insert middle | O(n) | O(1) |
| Delete last | O(1) | O(1) |
| Delete beginning | O(n) | O(1) |
| Delete middle | O(n) | O(1) |
| Reverse in-place | O(n) | O(1) |
| Min / Max | O(n) | O(1) |
| Sum / Product | O(n) | O(1) |
| Frequency counting | O(n) average | O(k) |
| Copy | O(n) | O(n) |
| 2D traversal | O(r × c) | O(1) |

### Complexity rules to remember

1. One loop over `n` elements → usually **O(n)**.
2. Two nested loops over `n` elements → usually **O(n²)**.
3. Repeatedly halving the search space → **O(log n)**.
4. Creating another array of size `n` → **O(n)** extra space.
5. Updating an array element by index → **O(1)**.
6. Inserting/deleting near the front or middle of a Python list → usually **O(n)** because of shifting.
